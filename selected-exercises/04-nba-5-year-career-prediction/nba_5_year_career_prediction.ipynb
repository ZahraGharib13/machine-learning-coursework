## feature engineering

```python
df["fgm"].value_counts()
```

```python
df["fg"].value_counts()
```

```python
x_model = x.drop(columns=["name"])
y = df["target_5yrs"]

x_train, x_val, y_train, y_val = train_test_split(x_model, y, test_size=0.2, random_state=42, stratify=y)
```

```python
def create_features(df):
    df = df.copy()

    # Playing time / role
    df["min_per_game"] = df["min"] / (df["gp"] + 1)

    # Scoring
    df["pts_per_min"] = df["pts"] / (df["min"] + 1)

    # Shooting efficiency
    df["fg_efficiency"] = df["fgm"] / (df["fga"] + 1)
    df["three_point_efficiency"] = df["3p_made"] / (df["3pa"] + 1)
    df["ft_efficiency"] = df["ftm"] / (df["fta"] + 1)

    # Playmaking
    df["ast_tov_ratio"] = df["ast"] / (df["tov"] + 1)

    # Rebounding contribution
    df["reb_per_min"] = df["reb"] / (df["min"] + 1)

    # Defensive contribution
    df["defensive_actions"] = df["stl"] + df["blk"]

    # All-around contribution
    df["all_around_score"] = (
        df["pts"] +
        df["reb"] +
        df["ast"] +
        df["stl"] +
        df["blk"])

    return df
```

```python
X_train_fe = create_features(x_train)
X_val_fe = create_features(x_val)
```

## only on the training data to avoid data leakage

```python
def add_threshold_features(train, val):

    train = train.copy()
    val = val.copy()

    cols = ["pts", "min", "ast", "reb", "stl", "blk"]

    for col in cols:
        threshold = train[col].quantile(0.75)

        train[f"high_{col}"] = (train[col] >= threshold).astype(int)
        val[f"high_{col}"] = (val[col] >= threshold).astype(int)

    return train, val
```

```python
X_train_fe, X_val_fe = add_threshold_features(X_train_fe, X_val_fe)
```

```python
print(X_train_fe.shape)
X_val_fe.shape
```

```python
def add_interaction_features(df):
    df = df.copy()

    # Offensive impact
    df["pts_per_shot"] = df["pts"] / (df["fga"] + 1)

    df["scoring_volume"] = df["pts"] * df["min"]

    # Playmaking quality
    df["assist_quality"] = df["ast"] / (df["tov"] + 1)

    # Rebounding role
    df["rebound_activity"] = df["reb"] * df["min"]

    # Two-way impact
    df["two_way_impact"] = (
        df["stl"] + df["blk"]
    ) * df["min"]

    return df
```

```python
X_train_fe = add_interaction_features(X_train_fe)

X_val_fe = add_interaction_features(X_val_fe)
```

```python
def add_shooting_features(df):
    df = df.copy()

    # 3 point volume adjusted by success
    df["three_point_attempt_success"] = (
        df["3p_made"] / (df["3pa"] + 1)
    )

    # total made shots efficiency
    df["made_shot_ratio"] = (
        df["fgm"] / (df["fga"] + 1)
    )

    # free throw reliability
    df["free_throw_volume_efficiency"] = (
        df["ftm"] * df["ft"]
    )

    return df
```

```python
X_train_fe = add_shooting_features(X_train_fe)
X_val_fe = add_shooting_features(X_val_fe)
```

```python
def add_role_features(train, val):
    train = train.copy()
    val = val.copy()

    train["minutes_per_game"] = train["min"] / (train["gp"] + 1)
    val["minutes_per_game"] = val["min"] / (val["gp"] + 1)

    train["games_minutes"] = train["gp"] * train["min"]
    val["games_minutes"] = val["gp"] * val["min"]

    minutes_median = train["min"].median()
    train["starter_role_proxy"] = (train["min"] >= minutes_median).astype(int)
    val["starter_role_proxy"] = (val["min"] >= minutes_median).astype(int)

    return train, val
```

```python
X_train_fe, X_val_fe = add_role_features(X_train_fe, X_val_fe)
```

```python
def add_defensive_features(df):
    df = df.copy()

    df["defensive_rebound_rate"] = (
        df["dreb"] / (df["reb"] + 1)
    )

    df["stocks"] = (
        df["stl"] + df["blk"]
    )

    return df
```

```python
X_train_fe = add_defensive_features(X_train_fe)
X_val_fe = add_defensive_features(X_val_fe)
```

## pipeline with feature engineering dataset

```python
feature_engineered_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000, random_state=42)),])

feature_engineered_pipeline.fit(X_train_fe, y_train)
```

```python
y_val_prob = feature_engineered_pipeline.predict_proba(X_val_fe)[:, 1]

auc = roc_auc_score(y_val, y_val_prob)

print("Feature Engineered ROC-AUC:", auc)
```

```python
coef = pd.DataFrame({
    "feature": X_train_fe.columns,
    "coefficient": feature_engineered_pipeline.named_steps["model"].coef_[0]
})

coef["abs_coef"] = coef["coefficient"].abs()

coef.sort_values(
    "abs_coef",
    ascending=False
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
import joblib

# Full labeled training data
X_full = df.drop(columns=["target_5yrs", "name"])
y_full = df["target_5yrs"]

# Unlabeled test data
X_test = test.drop(columns=["name"])

# Apply the same feature engineering used during validation
X_full_fe = create_features(X_full)
X_test_fe = create_features(X_test)

X_full_fe, X_test_fe = add_threshold_features(X_full_fe, X_test_fe)

X_full_fe = add_interaction_features(X_full_fe)
X_test_fe = add_interaction_features(X_test_fe)

X_full_fe = add_shooting_features(X_full_fe)
X_test_fe = add_shooting_features(X_test_fe)

X_full_fe, X_test_fe = add_role_features(X_full_fe, X_test_fe)

X_full_fe = add_defensive_features(X_full_fe)
X_test_fe = add_defensive_features(X_test_fe)

# Keep scaling and Logistic Regression together
final_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000, random_state=42)),
])

final_pipeline.fit(X_full_fe, y_full)
```

```python
test_pred = final_pipeline.predict_proba(X_test_fe)[:, 1]

submission = pd.DataFrame({
    "target_5yrs": test_pred
})

submission.head()
```

```python
from pathlib import Path
Path("outputs").mkdir(exist_ok=True)

joblib.dump(final_pipeline, "outputs/nba_5yr_model.joblib")
submission.to_csv("outputs/submission.csv", index=False)
```
